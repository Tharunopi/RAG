```python
from googlesearch import search
from utils.timer import timer
```

```python
@timer
def search_google(query:str, num_results:int=10):
    engine = search(query, num_results=10, unique=True)
    return [i for i in engine][1:]
```

```python
result, time_taken = search_google(query="what is elon musk age")
print(result)
print(time_taken)
```
